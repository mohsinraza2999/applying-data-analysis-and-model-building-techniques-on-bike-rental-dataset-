```python
import pandas as pd
from matplotlib import pyplot as plt
```

```python
data=pd.read_csv("SeoulBikeData.csv", encoding='unicode_escape')
```

```python
data.head()
```

```python
data.isnull().sum()
```

```python
duplicate_data=data.duplicated(keep=False)
print(duplicate_data.sum())
```

```python
data.describe()
```

```python
data.info()
```

```python
data['Functioning Day'] = data['Functioning Day'].replace(to_replace=['No', 'Yes'], value=[0, 1])
```

```python
data.info()
```

```python
data["Holiday"].value_counts()
```

```python
data['Holiday'] = data['Holiday'].replace(to_replace=['No Holiday', 'Holiday'], value=[0, 1])
```

```python
data["Seasons"].value_counts()
```

```python
data['Seasons'] = data['Seasons'].replace(to_replace=['Spring', 'Summer','Autumn','Winter'], value=[0, 1,2,3])
```

```python
data["Seasons"].value_counts()
```

```python
grouping = data.groupby('Seasons')["Rented Bike Count"].sum()
```

```python
grouping
```

```python
plt.plot(data[["Seasons"]],data["Rented Bike Count"])
```

```python
from scipy.stats import pearsonr


# Calculate correlation coefficient
correlation_coefficient, _ = pearsonr(data["Temperature(°C)"], data["Seasons"])

print("Correlation Coefficient (SciPy):", correlation_coefficient)
```

```python
df=pd.DataFrame(columns=[
    'Rented Bike Count', 'Hour', 'Temperature(°C)', 'Humidity(%)','Wind speed (m/s)', 
    'Visibility (10m)', 'Dew point temperature(°C)',
    'Solar Radiation (MJ/m2)', 'Rainfall(mm)', 'Snowfall (cm)', 'Seasons',
    'Holiday', 'Functioning Day'])
df=data[['Rented Bike Count', 'Hour', 'Temperature(°C)', 'Humidity(%)','Wind speed (m/s)', 
    'Visibility (10m)', 'Dew point temperature(°C)',
    'Solar Radiation (MJ/m2)', 'Rainfall(mm)', 'Snowfall (cm)', 'Seasons',
    'Holiday', 'Functioning Day']]

corr=df.corr(method = 'pearson',numeric_only=False)
corr
```

```python
import seaborn as sns
```

```python
plt.figure(figsize=(10,8), dpi =500)
sns.heatmap(corr,annot=True,fmt=".2f", linewidth=.5)
plt.show()
```
